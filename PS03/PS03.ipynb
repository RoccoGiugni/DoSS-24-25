# 1 - THEORETICAL STUDY OF THE SYSTEM
Brief review on the stability conditions and the engineering of the system.

**STABILITY CONDITIONS**

The gravitational instability of a disk is defined through the Toomre parameter:

$Q = \frac{2\sigma_{*}\Omega}{\pi G\Sigma}$

where $\sigma_{*}$ is the velocity dispersion, $\Omega$ the angular velocity and $\Sigma$ the surface mass density. In general, all these quantities are functions of the distance from the center of the disk (from here referred to just as 'center').

The disk is gravitationally stable for $Q > 1$, while it's expected to collapse and fragment in smaller clumps for $Q < 1$.

**ENGINEERING THE SYSTEM**

To keep things simple, we aim to build a disk where the stability condition stop holding above a certain distance from the center $d$. The expectetion is that p.cles inside $d$ will move on circular orbits, while p.cles in the outskirts will collapse into clumps.

Starting from the surface density, we choose:

$\Sigma = \frac{v_0^2}{2\pi G r}$

where $r$ is the distance from the center and $v_0$ is an arbitrary constant which name make sense in light of what follows. With this choice of $\Sigma$ the tangential velocity is $v_c = v_0$, thus the angular velocity is:

$\Omega = \frac{v_0}{r}$

The period of a p.cle at radius $r$ is:

$T = \frac{2\pi r}{v_0}$

Now the Toomre parameter end up being:

$Q = \frac{4\pi \sigma_*}{v_0}$

impling some sort of dependence on $r$ of $\sigma_*$ if the same is wanted for $Q$. We choose the velocity dispertion to be inversely proportional to the radius. Considering its dimesions, we define it as:

$\sigma_* = \frac{1}{4} \frac{Rv_0}{2\pi r}$

where $R$ is the radius of the disk. The resulting Toomre parameter is:

$Q = \frac{1}{2} \frac{R}{r}$
